```python
df = df.drop_duplicates().reset_index(drop=True)
```

```python
df.columns
```

```python
df = df[['title', 'tagline', 'overview', 'genres', 'vote_average','popularity']]
```

```python
df.isnull().sum()

```

```python
df = df.dropna(subset=['title'])
```

```python
df['overview'] = df['overview'].fillna('')
```

```python
import ast
df['genres'] = df['genres'].apply(lambda x: ' '.join([i['name'] for i in ast.literal_eval(x)]))
```

```python
df['tagline'] = df['tagline'].fillna('')
```

```python
df['vote_average'] = df['vote_average'].fillna(df['vote_average'].mean())
```

```python
df['popularity'] = df['popularity'].fillna(df['popularity'].mode())
```

```python
df['tag'] = df['overview'] + df['genres'] + df['tagline']
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
import re

def preprocessing(text):
  text = text.lower()
  text = re.sub(r'[^\w\s]', ' ', text)
  words = text.split()
  words = [lemmatizer.lemmatize(word) for word in words if word not in stop_words]
  return ' '.join(words)
```

```python
df['tag'][0]
```

```python
df['tag'] = df['tag'].apply(preprocessing)
```

```python
df['tag'][0]
```

```python
df.head()
```

```python
df = df.reset_index(drop=True)
```

```python
indices = pd.Series(df.index,index = df['title']).drop_duplicates()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(max_features=5000, ngram_range=(1,2), stop_words='english')
```

```python
tfidf_matrix = tfidf.fit_transform(df['tag'])
```

```python
tfidf_matrix.shape
```

```python
print(tfidf_matrix.toarray())
```

```python
from sklearn.metrics.pairwise import linear_kernel
```

```python
def recommend(title, n =10):
  if title not in indices :
    return ["Movie not found"]
  idx = indices[title]
  sim_score = linear_kernel(tfidf_matrix[idx],tfidf_matrix).flatten()
  sim_idx = sim_score.argsort()[::-1][1:n+1]

  return df['title'].iloc[sim_idx]
```

```python
recommend('Toy Story',10)
```

```python
import pickle
pickle.dump(tfidf_matrix,open('tfidf_matrix.pkl','wb'))

pickle.dump(tfidf,open('tfidf.pkl','wb'))

df.to_pickle('df.pkl')

pickle.dump(indices,open('indices.pkl','wb'))
```
